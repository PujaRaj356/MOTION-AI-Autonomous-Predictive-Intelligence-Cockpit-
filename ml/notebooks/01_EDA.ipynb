# Exploratory Data Analysis (EDA) - Predictive Maintenance

This notebook covers the initial data loading, cleaning, and exploratory data analysis (EDA) for the MOTION-AI Predictive Maintenance dataset.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Set plotting style
plt.style.use('ggplot')
sns.set_theme(style="whitegrid")
```

## 1. Data Loading
We will load the sensor telemetry data and failure logs here. Ensure your dataset (e.g., CMAPSS or AI4I 2020) is placed in the `data/` directory.

```python
# Example: Load the dataset (replace with your actual dataset path)
# df = pd.read_csv('../../data/sensor_telemetry.csv')

# For demonstration, we'll create a dummy dataframe matching our expected schema
df = pd.DataFrame({
    'machine_id': ['M-101', 'M-102', 'M-103', 'M-104'],
    'air_temperature': [24.5, 31.2, 22.1, 28.5],
    'process_temperature': [34.8, 42.5, 33.1, 39.0],
    'rpm': [1540, 2450, 1420, 2100],
    'torque': [42.1, 68.4, 40.5, 55.2],
    'tool_wear': [45, 215, 12, 120],
    'failure': [0, 1, 0, 0]
})

display(df.head())
```

## 2. Data Cleaning & Preprocessing
Handle missing values, encode categorical variables, and scale numerical features.

```python
# Check for missing values
print("Missing values:\n", df.isnull().sum())

# Example scaling (StandardScaler)
from sklearn.preprocessing import StandardScaler

features = ['air_temperature', 'process_temperature', 'rpm', 'torque', 'tool_wear']
scaler = StandardScaler()
df[features] = scaler.fit_transform(df[features])

display(df.head())
```

## 3. Exploratory Data Analysis (EDA)
Visualize the distribution of features and their correlation with machine failure.

```python
# Correlation Matrix
plt.figure(figsize=(8, 6))
corr = df.drop(columns=['machine_id']).corr()
sns.heatmap(corr, annot=True, cmap='coolwarm', fmt=".2f")
plt.title("Feature Correlation Matrix")
plt.show()
```
